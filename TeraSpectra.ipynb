```python
import numpy as np

# ENVI file specifications
lines = 1370
samples = 1024
bands = 216
header_offset = 3547136

# Read BIL formatted hyperspectral data
raw = np.memmap(
    "13_G.hyspex",
    dtype="<u2",
    mode="r",
    offset=header_offset,
    shape=(lines, bands, samples)
)

# Convert BIL layout to (lines, samples, bands)
img = raw.transpose(0, 2, 1)

print("Hyperspectral image loaded!")
print("Image shape:", img.shape)
print("Data type:", img.dtype)
print("Total spectral bands:", img.shape[2])
```

```python
import matplotlib.pyplot as plt
import numpy as np

band = np.asarray(img[:, :, 100])

# Ignore invalid pixels
valid_pixels = band != 4095

# Contrast adjustment
vmin, vmax = np.percentile(
    band[valid_pixels], [2, 98]
)

plt.figure(figsize=(14, 7))

plt.imshow(
    band,
    cmap="gray",
    vmin=vmin,
    vmax=vmax
)

plt.title("TerraSpectra - Complete VNIR Band 100")
plt.xlabel("Image Width")
plt.ylabel("Image Height")
plt.colorbar(label="Pixel Intensity")

plt.show()

print("Image shape:", img.shape)
print("Valid pixels:", valid_pixels.sum())
print("Ignored pixels:", (~valid_pixels).sum())
```

```python
import numpy as np
import matplotlib.pyplot as plt

band = np.asarray(img[:, :, 100], dtype=np.float32)

# Exclude invalid pixels
valid = band != 4095

# Calculate average intensity for each row
row_means = np.array([
    np.mean(band[i][valid[i]])
    if valid[i].any() else np.nan
    for i in range(band.shape[0])
])

plt.figure(figsize=(14, 5))

plt.plot(row_means)
plt.title("VNIR Band 100 - Row Intensity Profile")
plt.xlabel("Image Row")
plt.ylabel("Average Pixel Intensity")
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Select the suspected leaf region
leaf_area = np.asarray(img[520:1030, :, 100])

# Display the cropped image
plt.figure(figsize=(14, 6))

plt.imshow(
    leaf_area,
    cmap="gray",
    vmin=0,
    vmax=5000
)

plt.title("TerraSpectra - Suspected Wheat Leaf Regions")
plt.xlabel("Image Width")
plt.ylabel("Cropped Image Height")
plt.colorbar(label="Pixel Intensity")

plt.show()

print("Cropped image shape:", leaf_area.shape)
```

```python
from skimage.filters import threshold_otsu
import numpy as np
import matplotlib.pyplot as plt

# Select the leaf inspection region
roi = np.asarray(img[520:1030, :, 100])

# Exclude invalid pixels
valid_pixels = roi[roi != 4095]

# Calculate automatic threshold
threshold = threshold_otsu(valid_pixels)

print("Otsu Threshold:", threshold)

print("Minimum intensity:", valid_pixels.min())
print("Maximum intensity:", valid_pixels.max())
print("Mean intensity:", round(valid_pixels.mean(), 2))
```

```python
# Extract Band 100 from the ROI
roi = np.asarray(img[520:1030, :, 100])

# Identify valid pixels
valid_mask = roi != 4095

# Create binary mask
leaf_mask = (roi > threshold) & valid_mask

# Visualize the mask
plt.figure(figsize=(14, 6))

plt.imshow(leaf_mask, cmap="gray")

plt.title("TerraSpectra - Initial Binary Leaf Mask")
plt.xlabel("Image Width")
plt.ylabel("Image Height")

plt.show()

print("Mask shape:", leaf_mask.shape)
print("Potential foreground pixels:", leaf_mask.sum())
print("Background pixels:", (~leaf_mask).sum())
```

```python
from scipy import ndimage
import numpy as np

# Find connected foreground regions
labeled_mask, num_components = ndimage.label(leaf_mask)

# Calculate component sizes
component_sizes = np.bincount(
    labeled_mask.ravel()
)[1:]

# Sort component sizes
sorted_sizes = np.sort(component_sizes)[::-1]

print("Total connected components:", num_components)

print("\nLargest 10 components:")

for i, size in enumerate(sorted_sizes[:10]):
    print(f"Component {i+1}: {size} pixels")
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import ndimage

# Find component positions
component_objects = ndimage.find_objects(labeled_mask)

# Get five largest component labels
top_labels = np.argsort(component_sizes)[::-1][:5] + 1

plt.figure(figsize=(14, 7))

plt.imshow(roi, cmap="gray", vmin=0, vmax=5000)

ax = plt.gca()

for label in top_labels:

    region = component_objects[label - 1]

    if region is None:
        continue

    y_slice, x_slice = region

    height = y_slice.stop - y_slice.start
    width = x_slice.stop - x_slice.start

    rect = plt.Rectangle(
        (x_slice.start, y_slice.start),
        width,
        height,
        fill=False,
        edgecolor="red",
        linewidth=2
    )

    ax.add_patch(rect)

    plt.text(
        x_slice.start,
        y_slice.start - 5,
        f"Component {label}",
        color="yellow",
        fontsize=9
    )

plt.title("TerraSpectra - Largest Connected Components")
plt.xlabel("Image Width")
plt.ylabel("Image Height")
plt.show()
```

```python
from scipy import ndimage
import numpy as np

# Calculate component properties
objects = ndimage.find_objects(labeled_mask)

top_labels = np.argsort(component_sizes)[::-1][:10] + 1

print("COMPONENT DETAILS")
print("-" * 50)

for label in top_labels:

    region = objects[label - 1]

    if region is None:
        continue

    y_slice, x_slice = region

    height = y_slice.stop - y_slice.start
    width = x_slice.stop - x_slice.start

    y_center = (y_slice.start + y_slice.stop) // 2
    x_center = (x_slice.start + x_slice.stop) // 2

    print(f"\nComponent: {label}")
    print("Pixels:", component_sizes[label - 1])
    print("Bounding box height:", height)
    print("Bounding box width:", width)
    print("Center:", (y_center, x_center))
```

```python
from scipy import ndimage
import numpy as np
import matplotlib.pyplot as plt

# Create a small 3x3 kernel
kernel = np.ones((3, 3), dtype=bool)

# Apply morphological closing
refined_mask = ndimage.binary_closing(
    leaf_mask,
    structure=kernel
)

# Count new connected components
refined_labels, refined_count = ndimage.label(refined_mask)

print("Original components:", num_components)
print("Refined components:", refined_count)

# Visualize refined mask
plt.figure(figsize=(14, 6))

plt.imshow(refined_mask, cmap="gray")

plt.title("TerraSpectra - Refined Leaf Mask")
plt.xlabel("Image Width")
plt.ylabel("Image Height")

plt.show()
```

```python
# Calculate refined component sizes
refined_sizes = np.bincount(
    refined_labels.ravel()
)[1:]

# Sort component sizes
sorted_refined_sizes = np.sort(
    refined_sizes
)[::-1]

print("REFINED COMPONENT ANALYSIS")
print("-" * 40)

print("Total components:", refined_count)

for i, size in enumerate(sorted_refined_sizes[:10]):
    print(f"Component {i+1}: {size} pixels")
```

```python
# Find the three largest component labels
top3_labels = np.argsort(refined_sizes)[::-1][:3] + 1

# Create a separate mask for inspection
top3_mask = np.isin(refined_labels, top3_labels)

plt.figure(figsize=(14, 6))

plt.imshow(top3_mask, cmap="gray")

plt.title("TerraSpectra - Three Largest Foreground Regions")
plt.xlabel("Image Width")
plt.ylabel("Image Height")

plt.show()

print("Selected component labels:", top3_labels)
print("Selected pixels:", top3_mask.sum())
```

```python
plt.figure(figsize=(15, 7))

# Display original Band 100
plt.imshow(roi, cmap="gray", vmin=0, vmax=5000)

# Overlay segmentation boundaries
plt.contour(
    top3_mask,
    levels=[0.5],
    colors="red",
    linewidths=1
)

plt.title("TerraSpectra - Leaf Segmentation Boundary Validation")
plt.xlabel("Image Width")
plt.ylabel("Image Height")

plt.show()
```

```python
# Extract the original hyperspectral ROI
roi_cube = img[520:1030, :, :]

# Extract spectral information for selected leaf pixels
leaf_spectra = roi_cube[top3_mask]

# Preserve pixel coordinates
leaf_coords = np.argwhere(top3_mask)

# Convert cropped coordinates to original image coordinates
leaf_coords[:, 0] += 520

print("HYPERSPECTRAL LEAF EXTRACTION")
print("-" * 40)

print("Original cube shape:", img.shape)
print("ROI cube shape:", roi_cube.shape)

print("Extracted leaf spectra shape:", leaf_spectra.shape)

print("Total selected pixels:", len(leaf_spectra))
print("Spectral bands per pixel:", leaf_spectra.shape[1])

# Check invalid values
invalid_count = np.sum(leaf_spectra == 4095)

print("Invalid values (4095):", invalid_count)
```

```python
# Identify invalid spectral values
invalid_mask = (leaf_spectra == 4095)

# Count invalid values per pixel
invalid_pixels = invalid_mask.any(axis=1)

# Count invalid values per spectral band
invalid_per_band = invalid_mask.sum(axis=0)

print("INVALID DATA ANALYSIS")
print("-" * 40)

print("Total invalid measurements:",
      invalid_mask.sum())

print("Pixels containing invalid values:",
      invalid_pixels.sum())

print("Pixels without invalid values:",
      (~invalid_pixels).sum())

print("Bands containing invalid values:",
      np.sum(invalid_per_band > 0))

print("Maximum invalid values in one band:",
      invalid_per_band.max())
```

```python
# Select pixels without invalid measurements
valid_pixel_mask = ~invalid_pixels

# Extract completely valid spectra
clean_leaf_spectra = leaf_spectra[valid_pixel_mask]

# Preserve corresponding pixel coordinates
clean_leaf_coords = leaf_coords[valid_pixel_mask]

print("CLEAN HYPERSPECTRAL DATA")
print("-" * 40)

print("Original spectra:", leaf_spectra.shape)

print("Clean spectra:", clean_leaf_spectra.shape)

print("Clean coordinates:", clean_leaf_coords.shape)

print("Remaining invalid values:",
      np.sum(clean_leaf_spectra == 4095))

print("Removed pixels:", invalid_pixels.sum())
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Calculate average intensity for each band
mean_spectrum = np.mean(clean_leaf_spectra, axis=0)

# Calculate standard deviation
std_spectrum = np.std(clean_leaf_spectra, axis=0)

# Plot average spectral signature
plt.figure(figsize=(12, 5))

plt.plot(mean_spectrum)

plt.title("TerraSpectra - Average Leaf Spectral Signature")
plt.xlabel("Spectral Band")
plt.ylabel("Average Pixel Intensity")

plt.grid(True)
plt.show()

print("Number of spectral bands:", len(mean_spectrum))
print("Minimum mean intensity:", round(mean_spectrum.min(), 2))
print("Maximum mean intensity:", round(mean_spectrum.max(), 2))
```

```python
plt.figure(figsize=(12, 5))

bands = np.arange(216)

plt.plot(
    bands,
    mean_spectrum,
    label="Mean Spectrum"
)

plt.fill_between(
    bands,
    mean_spectrum - std_spectrum,
    mean_spectrum + std_spectrum,
    alpha=0.2,
    label="Standard Deviation"
)

plt.title("TerraSpectra - Spectral Variation")
plt.xlabel("Spectral Band")
plt.ylabel("Pixel Intensity")

plt.legend()
plt.grid(True)
plt.show()
```
