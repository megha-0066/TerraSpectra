```python
import numpy as np
import matplotlib.pyplot as plt
import os

try:
    # Calculate how many rows are actually present in the truncated file on disk
    bytes_per_row = img.shape[1] * img.shape[2] * np.dtype(img.dtype).itemsize
    actual_file_size = os.path.getsize("/content/13_G.hyspex")

    # The maximum row index we can read without hitting an EOFError
    max_safe_row = int(actual_file_size // bytes_per_row)

    print("TERRASPECTRA - SAFE ROI RECOVERY")
    print("-" * 40)
    print(f"Original height in metadata: {img.shape[0]} rows")
    print(f"Max safe row index on disk: {max_safe_row} rows")

    # Slice within safe limits (using a 10-row buffer to prevent exact boundary reading errors)
    safe_end_row = min(1040, max_safe_row - 10)
    safe_start_row = 0

    print(f"Slicing rows {safe_start_row} to {safe_end_row}...")

    # Slicing the 'img' object directly returns the safe chunk as a NumPy array
    roi_cube = img[safe_start_row:safe_end_row, :, :]
    roi_image = roi_cube[:, :, 100]

    print("ROI cube shape:", roi_cube.shape)
    print("ROI image shape:", roi_image.shape)

    plt.figure(figsize=(12, 6))
    plt.imshow(roi_image, cmap="gray")
    plt.title(f"TerraSpectra - Reconstructed Leaf (Slices {safe_start_row}:{safe_end_row})")
    plt.xlabel("Image Width")
    plt.ylabel("Image Height")
    plt.colorbar(label="Pixel Intensity")
    plt.show()
except Exception as e:
    print(f"An error occurred while reading the ROI slice from the disk: {e}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import os
from skimage.filters import threshold_otsu

# Ensure roi_cube exists; if not, recreate it safely
if 'roi_cube' not in globals():
    print("roi_cube not found in memory. Re-extracting from img...")
    bytes_per_row = img.shape[1] * img.shape[2] * np.dtype(img.dtype).itemsize
    actual_file_size = os.path.getsize("/content/13_G.hyspex")
    max_safe_row = int(actual_file_size // bytes_per_row)
    safe_end_row = min(1040, max_safe_row - 10)
    roi_cube = img[0:safe_end_row, :, :]

# Select spectral band 100
band_image = roi_cube[:, :, 100]

# Remove invalid pixels
valid_pixels = band_image[band_image != 4095]

# Calculate Otsu threshold
threshold = threshold_otsu(valid_pixels)

# Create binary leaf mask
leaf_mask = (band_image > threshold) & (band_image != 4095)

print("TERRASPECTRA - LEAF SEGMENTATION")
print("-" * 40)
print("Otsu threshold:", threshold)
print("Selected pixels:", leaf_mask.sum())
print("Mask shape:", leaf_mask.shape)

# Display mask
plt.figure(figsize=(12, 6))
plt.imshow(leaf_mask, cmap="gray")
plt.title("TerraSpectra - Reconstructed Leaf Mask")
plt.xlabel("Image Width")
plt.ylabel("Image Height")
plt.show()
```

```python
from skimage.measure import label
from skimage.morphology import remove_small_objects

# Identify connected regions
components = label(leaf_mask)

# Remove small isolated regions
refined_mask = remove_small_objects(
    components,
    min_size=100
)

# Convert to binary mask
refined_mask = refined_mask > 0

print("TERRASPECTRA - MASK REFINEMENT")
print("-" * 40)

print("Original pixels:", leaf_mask.sum())
print("Refined pixels:", refined_mask.sum())
print("Removed pixels:", leaf_mask.sum() - refined_mask.sum())

plt.figure(figsize=(12, 6))
plt.imshow(refined_mask, cmap="gray")
plt.title("TerraSpectra - Refined Leaf Mask")
plt.xlabel("Image Width")
plt.ylabel("Image Height")
plt.show()
```

```python

import os
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

print("TERRASPECTRA - DAY 6: DATASET SPLITTING")
print("=" * 50)

# 1. Biological Grouping (Group by Pot, Side, Leaf to find unique independent biological samples)
df_leaves = df.copy()
df_leaves['Leaf_ID_Unique'] = df_leaves.apply(lambda r: f"Pot{int(r['Pot [1]'])}_Side{r['Side [2]']}_Leaf{int(r['Leaf [3]'])}", axis=1)

unique_leaves = df_leaves.drop_duplicates(subset=['Leaf_ID_Unique']).copy()
total_biological_samples = len(unique_leaves)

print(f"Total unique biological leaf samples found: {total_biological_samples}")

# 2. Extract Distributions
print("\n1. LABEL & CLASS DISTRIBUTION (BIOLOGICAL SAMPLES)")
print("-" * 50)
treatment_dist = unique_leaves['Treatment [5]'].value_counts()
genotype_dist = unique_leaves['Genotype [4]'].value_counts()
confidence_dist = unique_leaves['Confidence level [6]'].value_counts()

print("Treatment Distribution:\n", treatment_dist)
print("\nGenotype Distribution:\n", genotype_dist)
print("\nConfidence Level Distribution:\n", confidence_dist)

# 3. Create Splits (70% Train, 15% Val, 15% Test) ensuring no Leaf ID leaks
# Stratify by Genotype & Treatment combined to maintain balanced splits
unique_leaves['Stratify_Group'] = unique_leaves['Genotype [4]'].astype(str) + "_" + unique_leaves['Treatment [5]'].astype(str)

# First split: Train (70%) and Temp (30%)
train_leaves, temp_leaves = train_test_split(
    unique_leaves,
    test_size=0.30,
    random_state=42,
    stratify=unique_leaves['Stratify_Group']
)

# Second split: Val (15%) and Test (15%)
val_leaves, test_leaves = train_test_split(
    temp_leaves,
    test_size=0.50,
    random_state=42,
    stratify=temp_leaves['Stratify_Group']
)

# Map back to pixel level / complete leaf descriptions
train_set = df_leaves[df_leaves['Leaf_ID_Unique'].isin(train_leaves['Leaf_ID_Unique'])]
val_set = df_leaves[df_leaves['Leaf_ID_Unique'].isin(val_leaves['Leaf_ID_Unique'])]
test_set = df_leaves[df_leaves['Leaf_ID_Unique'].isin(test_leaves['Leaf_ID_Unique'])]

print("\n2. SPLIT RESULTS (LEAF LEVEL)")
print("-" * 50)
print(f"Training set size: {len(train_set)} leaves ({len(train_set)/total_biological_samples*100:.1f}%)")
print(f"Validation set size: {len(val_set)} leaves ({len(val_set)/total_biological_samples*100:.1f}%)")
print(f"Testing set size: {len(test_set)} leaves ({len(test_set)/total_biological_samples*100:.1f}%)")

# 4. Leakage Verification
overlap = set(train_set['Leaf_ID_Unique']).intersection(set(test_set['Leaf_ID_Unique']))
print(f"\nOverlap check between Train and Test sets: {len(overlap)} overlapping leaf IDs (Leakage safe!)" if not overlap else "\n[WARNING] DATA LEAKAGE DETECTED!")

# 5. Save Splits
os.makedirs("/content/terraspectra_day6", exist_ok=True)
train_set.to_csv("/content/terraspectra_day6/train_split.csv", index=False)
val_set.to_csv("/content/terraspectra_day6/val_split.csv", index=False)
test_set.to_csv("/content/terraspectra_day6/test_split.csv", index=False)

print("\nAll split files successfully saved in: /content/terraspectra_day6/")
```

```python
import os
import pandas as pd
import numpy as np
import spectral

print("============================================================")
print("TERRASPECTRA: TASK 1 - COMPREHENSIVE DATASET FILE SEARCH")
print("============================================================")

search_dirs = ['/content']
gdrive_path = '/content/drive/MyDrive'
if os.path.exists(gdrive_path):
    search_dirs.append(gdrive_path)
else:
    print("[Notice] Google Drive (/content/drive/MyDrive) is not mounted.")

target_extensions = (".hyspex", ".hdr", ".raw", ".npy", ".npz", ".mat", ".tif", ".tiff", ".csv", ".xlsx")
found_files = []

for directory in search_dirs:
    print(f"Searching in: {directory}...")
    for root, dirs, files in os.walk(directory):
        for file in files:
            if file.lower().endswith(target_extensions):
                full_path = os.path.join(root, file)
                found_files.append({
                    'Filename': file,
                    'Path': full_path,
                    'Size_MB': round(os.path.getsize(full_path) / (1024 * 1024), 3)
                })

files_df = pd.DataFrame(found_files)
if not files_df.empty:
    display(files_df)
else:
    print("No relevant files found.")
```

```python
print("============================================================")
print("TERRASPECTRA: TASK 2 - HYPERSPECTRAL IMAGE METADATA IDENTIFICATION")
print("============================================================")

# Filter found files for image files and attempt to open them
hdr_files = [f for f in found_files if f['Filename'].lower().endswith('.hdr')]

for hdr_item in hdr_files:
    hdr_path = hdr_item['Path']
    base, _ = os.path.splitext(hdr_path)

    # Look for matching image binary
    img_path = None
    for ext in ['.hyspex', '.raw', '']:
        if os.path.exists(base + ext) and (base + ext) != hdr_path:
            img_path = base + ext
            break

    print(f"\nHDR File: {hdr_path}")
    print(f"Image Binary File: {img_path if img_path else 'None found'}")

    if img_path:
        try:
            img_obj = spectral.open_image(hdr_path)
            print(f"  - Can be opened: YES")
            print(f"  - Dimensions: {img_obj.shape[0]} rows x {img_obj.shape[1]} columns")
            print(f"  - Spectral Bands: {img_obj.shape[2]}")
        except Exception as e:
            print(f"  - Can be opened: NO (Error: {e})")
    else:
        print("  - Can be opened: NO (No associated binary raw file)")
```

```python
print("============================================================")
print("TERRASPECTRA: TASK 3 & 4 - ANNOTATION MATCHING AND QUALITY REPORT")
print("============================================================")

# Ensure clean df annotations are matched
df_report = df.copy()

# Identify matching pots on disk
available_pots_sides = []
for f in found_files:
    name, ext = os.path.splitext(f['Filename'])
    if ext.lower() == '.hyspex':
        parts = name.split('_')
        if len(parts) == 2:
            try:
                pot_num = int(parts[0])
                side_char = parts[1]
                available_pots_sides.append((pot_num, side_char))
            except ValueError:
                continue

def check_annot_match(row):
    try:
        row_pot = int(row['Pot [1]'])
        row_side = str(row['Side [2]']).strip()
        return (row_pot, row_side) in available_pots_sides
    except:
        return False

df_report['Has_Physical_Image'] = df_report.apply(check_annot_match, axis=1)
matched_annots = df_report[df_report['Has_Physical_Image']]

print(f"Total labels in 'df': {len(df_report)}")
print(f"Labels matched to physical hyperspectral files: {len(matched_annots)}")
if len(matched_annots) > 0:
    display(matched_annots)
else:
    print("No annotations matched to physical images.")

print("\n============================================================")
print("DATASET RECOVERY SUMMARY & SUITABILITY STATUS")
print("============================================================")
if len(matched_annots) < 10:
    print("STATUS: NOT READY FOR MODEL TRAINING")
    print("Reason: Insufficient matched biological samples available on disk for training a supervised classifier (RF/SVM).")
else:
    print("STATUS: READY")
```
