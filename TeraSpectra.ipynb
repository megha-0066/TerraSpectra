```python
!pip -q install remotezip

from remotezip import RemoteZip
import os

url = "https://entrepot.recherche.data.gouv.fr/api/access/datafile/:persistentId?persistentId=doi:10.57745/RKRD2T"

folder = "/content/drive/MyDrive/TerraSpectra"

files = [
    "20210331/13_G/13_G.hyspex",
    "20210331/13_G/13_G.hdr"
]

with RemoteZip(url) as z:
    for file in files:
        filename = os.path.basename(file)
        output = os.path.join(folder, filename)

        with z.open(file) as source, open(output, "wb") as target:
            while True:
                chunk = source.read(8 * 1024 * 1024)
                if not chunk:
                    break
                target.write(chunk)

        print("Downloaded:", filename)
```

```python
import numpy as np

row = 500
col = 500

spectrum = img.read_pixel(row, col)

print("Pixel location:", row, col)
print("Number of bands:", len(spectrum))
print("First 10 spectral values:", spectrum[:10])

print("Minimum:", np.min(spectrum))
print("Maximum:", np.max(spectrum))
print("Mean:", np.mean(spectrum))
```
