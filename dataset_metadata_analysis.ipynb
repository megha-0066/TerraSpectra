```python
import pandas
file_path = 'Data_description.csv'
df_raw = pd.read_csv(file_path, sep=';', encoding='cp1252', skiprows=2)
desc_df = df_raw.iloc[:, :6].copy()
desc_df.columns = ['Pot', 'Side', 'Leaf', 'Genotype', 'Treatment', 'Confidence_level']
desc_df = desc_df.dropna(subset=['Pot'])
print("=== 1. Loading and Inspecting Data Description ===")
print(f"Total valid samples: {len(desc_df)}")
print("\nFirst 3 rows:")
display(desc_df.head(3))

print("\nUnique Genotypes:")
print(desc_df['Genotype'].unique())

print("\nTreatments Summary:")
print(desc_df['Treatment'].value_counts())

print("\n---  Validation Executed Successfully without Errors! ---")
```
