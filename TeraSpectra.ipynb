```python
import pandas as pd

vnir_df = pd.read_csv(
    "VNIR_data.csv",
    sep=";",
    encoding="cp1252",
    header=[2, 3]
)

swir_df = pd.read_csv(
    "SWIR_data.csv",
    sep=";",
    encoding="cp1252",
    header=[2, 3]
)

print("VNIR shape:", vnir_df.shape)
print("SWIR shape:", swir_df.shape)

display(vnir_df.head())
display(swir_df.head())
```

```python
print("VNIR missing values:", vnir_df.isnull().sum().sum())
print("SWIR missing values:", swir_df.isnull().sum().sum())

print("\nVNIR treatment counts:")
print(vnir_df.iloc[:, 2].value_counts())

print("\nSWIR treatment counts:")
print(swir_df.iloc[:, 2].value_counts())
```

```python
treatment_col = vnir_df.columns[2]

print("Treatment column:", treatment_col)

print("\nTreatment counts including missing:")
print(vnir_df[treatment_col].value_counts(dropna=False))

print("\nRows with missing treatment:")
display(vnir_df[vnir_df[treatment_col].isna()])
```

```python
# Create clean copies using numeric Pot IDs only
vnir_clean = vnir_df[
    pd.to_numeric(vnir_df.iloc[:, 0], errors="coerce").notna()
].copy()

swir_clean = swir_df[
    pd.to_numeric(swir_df.iloc[:, 0], errors="coerce").notna()
].copy()

print("VNIR cleaned shape:", vnir_clean.shape)
print("SWIR cleaned shape:", swir_clean.shape)

print("\nVNIR treatment counts:")
print(vnir_clean.iloc[:, 2].value_counts())

print("\nSWIR treatment counts:")
print(swir_clean.iloc[:, 2].value_counts())
```

```python
print("VNIR observation columns:\n")

for i, col in enumerate(vnir_clean.columns[3:], start=3):
    print(i, col)
```

```python
date_summary = []

current_date = None

for i, col in enumerate(vnir_clean.columns[3:], start=3):

    header = str(col[0])
    side = str(col[1])

    if not header.startswith("Unnamed"):
        current_date = header

    values = pd.to_numeric(vnir_clean.iloc[:, i], errors="coerce")

    date_summary.append({
        "Date_DPI": current_date,
        "Side": side,
        "Total_Leaves": int(values.sum()),
        "Available_Records": int(values.notna().sum())
    })

date_summary_df = pd.DataFrame(date_summary)

display(date_summary_df)
```

```python
display(
    date_summary_df.pivot(
        index="Date_DPI",
        columns="Side",
        values="Available_Records"
    ).fillna(0).astype(int)
)
```

```python
import numpy as np
import os

# Image information from header
samples = 1024
bands = 216
header_offset = 3547136

# Available complete lines
file_size = os.path.getsize("13_G.hyspex")
bytes_per_line = samples * bands * 2

available_lines = (
    file_size - header_offset
) // bytes_per_line

print("Available lines:", available_lines)

# Read available complete image lines
raw = np.memmap(
    "13_G.hyspex",
    dtype=np.uint16,
    mode="r",
    offset=header_offset,
    shape=(available_lines, bands, samples)
)

# Convert BIL to (lines, samples, bands)
img = np.transpose(raw, (0, 2, 1))

print("\nImage loaded successfully!")
print("Image shape:", img.shape)
print("Data type:", img.dtype)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Select three spectral bands
rgb = img[:, :, [150, 100, 50]].astype(np.float32)

# Contrast adjustment
for i in range(3):
    low, high = np.percentile(rgb[:, :, i], [2, 98])
    rgb[:, :, i] = np.clip(
        (rgb[:, :, i] - low) / (high - low), 0, 1
    )

plt.figure(figsize=(12, 6))
plt.imshow(rgb)
plt.title("TerraSpectra - Partial VNIR False Colour Image")
plt.axis("off")
plt.show()
```
