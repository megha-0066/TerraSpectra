```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

# Upload VNIR file
uploaded = files.upload()
filename = next(iter(uploaded))

# Read VNIR dataset
df = pd.read_csv(
    filename,
    sep=";",
    skiprows=3,
    encoding="latin1",
    na_values=["NA", ""]
)

# Rename first three columns safely
df.columns = df.columns.astype(str).str.strip()
df.rename(columns={
    df.columns[0]: "Pot",
    df.columns[1]: "Genotype",
    df.columns[2]: "Treatment"
}, inplace=True)

# Treatment Distribution
treatment = df["Treatment"].dropna().value_counts()

print("VNIR Treatment Distribution")
print(treatment)

plt.figure(figsize=(8, 5))
treatment.plot(kind="bar")
plt.title("VNIR Treatment Distribution")
plt.xlabel("Treatment")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("VNIR_Treatment_Distribution.png", dpi=300)
plt.show()

# Genotype Distribution
genotype = df["Genotype"].dropna().value_counts()

print("VNIR Genotype Distribution")
print(genotype)

plt.figure(figsize=(8, 5))
genotype.plot(kind="bar")
plt.title("VNIR Genotype Distribution")
plt.xlabel("Genotype")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("VNIR_Genotype_Distribution.png", dpi=300)
plt.show()

print("Both VNIR graphs generated successfully!")

# Download graphs
files.download("VNIR_Treatment_Distribution.png")
files.download("VNIR_Genotype_Distribution.png")
```
