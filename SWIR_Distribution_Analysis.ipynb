```python


import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

# 1. Upload SWIR dataset
print("Upload SWIR_data.csv")
uploaded = files.upload()
filename = next(iter(uploaded))

# 2. Read SWIR dataset
df = pd.read_csv(
    filename,
    sep=";",
    skiprows=3,
    encoding="latin1",
    na_values=["NA", ""]
)

# Rename metadata columns
df.columns = df.columns.astype(str).str.strip()

df.rename(columns={
    df.columns[0]: "Pot",
    df.columns[1]: "Genotype",
    df.columns[2]: "Treatment"
}, inplace=True)

print("SWIR Dataset Shape:", df.shape)

# 3. SWIR Treatment Distribution
treatment = df["Treatment"].dropna().value_counts()

print("\nSWIR Treatment Distribution:")
print(treatment)

plt.figure(figsize=(8, 5))
treatment.plot(kind="bar")
plt.title("SWIR Treatment Distribution")
plt.xlabel("Treatment")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("SWIR_Treatment_Distribution.png", dpi=300)
plt.show()

# 4. SWIR Genotype Distribution
genotype = df["Genotype"].dropna().value_counts()

print("\nSWIR Genotype Distribution:")
print(genotype)

plt.figure(figsize=(8, 5))
genotype.plot(kind="bar")
plt.title("SWIR Genotype Distribution")
plt.xlabel("Genotype")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("SWIR_Genotype_Distribution.png", dpi=300)
plt.show()

print("\nSWIR ANALYSIS COMPLETED SUCCESSFULLY!")
```
