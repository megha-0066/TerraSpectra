```python
# Reconstruct the leaf region
roi_cube = cube[530:1040, :, :]

# Display spectral band 100
roi_image = roi_cube[:, :, 100]

print("TERRASPECTRA - ROI RECOVERY")
print("-" * 40)

print("Original cube:", cube.shape)
print("ROI cube:", roi_cube.shape)
print("ROI image:", roi_image.shape)

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
plt.imshow(roi_image, cmap="gray")
plt.title("TerraSpectra - Reconstructed Leaf ROI")
plt.xlabel("Image Width")
plt.ylabel("Image Height")
plt.colorbar(label="Pixel Intensity")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage.filters import threshold_otsu

# Select spectral band 100
band_image = roi_cube[:, :, 100]

# Remove invalid pixels
valid_pixels = band_image[band_image != 4095]

# Calculate Otsu threshold
threshold = threshold_otsu(valid_pixels)

# Create binary leaf mask
leaf_mask = (band_image > threshold) & (band_image != 4095)

print("TERRASPECTRA - LEAF SEGMENTATION")
print("-" * 40)

print("Otsu threshold:", threshold)
print("Selected pixels:", leaf_mask.sum())
print("Mask shape:", leaf_mask.shape)

# Display mask
plt.figure(figsize=(12, 6))
plt.imshow(leaf_mask, cmap="gray")
plt.title("TerraSpectra - Reconstructed Leaf Mask")
plt.xlabel("Image Width")
plt.ylabel("Image Height")
plt.show()
```

```python
from skimage.measure import label
from skimage.morphology import remove_small_objects

# Identify connected regions
components = label(leaf_mask)

# Remove small isolated regions
refined_mask = remove_small_objects(
    components,
    min_size=100
)

# Convert to binary mask
refined_mask = refined_mask > 0

print("TERRASPECTRA - MASK REFINEMENT")
print("-" * 40)

print("Original pixels:", leaf_mask.sum())
print("Refined pixels:", refined_mask.sum())
print("Removed pixels:", leaf_mask.sum() - refined_mask.sum())

plt.figure(figsize=(12, 6))
plt.imshow(refined_mask, cmap="gray")
plt.title("TerraSpectra - Refined Leaf Mask")
plt.xlabel("Image Width")
plt.ylabel("Image Height")
plt.show()
```

```python

import os
import pandas as pd
from sklearn.model_selection import GroupShuffleSplit

print("TERRASPECTRA - DAY 6")
print("POT-LEVEL DATASET SPLITTING")
print("=" * 50)

# Use the existing biological annotation dataset
data = df.copy()

pot_col = "Pot [1]"
label_col = "Treatment [5]"

# Ensure pot IDs are consistent
data[pot_col] = data[pot_col].astype(str)

# First split: 70% training, 30% temporary
split1 = GroupShuffleSplit(
    n_splits=1,
    test_size=0.30,
    random_state=42
)

train_idx, temp_idx = next(
    split1.split(data, groups=data[pot_col])
)

train_df = data.iloc[train_idx].copy()
temp_df = data.iloc[temp_idx].copy()

# Second split: Divide remaining data equally
split2 = GroupShuffleSplit(
    n_splits=1,
    test_size=0.50,
    random_state=42
)

val_idx, test_idx = next(
    split2.split(temp_df, groups=temp_df[pot_col])
)

val_df = temp_df.iloc[val_idx].copy()
test_df = temp_df.iloc[test_idx].copy()

# Check pot-level leakage
train_pots = set(train_df[pot_col])
val_pots = set(val_df[pot_col])
test_pots = set(test_df[pot_col])

assert train_pots.isdisjoint(val_pots)
assert train_pots.isdisjoint(test_pots)
assert val_pots.isdisjoint(test_pots)

# Display results
print("\nSPLIT RESULTS")

for name, subset in [
    ("Training", train_df),
    ("Validation", val_df),
    ("Testing", test_df)
]:
    print(f"\n{name}:")
    print("Records:", len(subset))
    print("Unique pots:", subset[pot_col].nunique())
    print("Treatment distribution:")
    print(subset[label_col].value_counts())

# Save separately without overwriting previous splits
output_dir = "/content/terraspectra_day6"
os.makedirs(output_dir, exist_ok=True)

train_df.to_csv(
    f"{output_dir}/train_annotations_pot_grouped.csv",
    index=False
)

val_df.to_csv(
    f"{output_dir}/val_annotations_pot_grouped.csv",
    index=False
)

test_df.to_csv(
    f"{output_dir}/test_annotations_pot_grouped.csv",
    index=False
)

print("\nPOT OVERLAP:",
      train_pots & val_pots |
      train_pots & test_pots |
      val_pots & test_pots)

print("\nDay 6 annotation split completed.")
```

```python
import torch
from torch.utils.data import Dataset, DataLoader
import os
import numpy as np
import matplotlib.pyplot as plt

# Create Day 8 Output Directory
output_dir_day8 = "/content/terraspectra_day8"
os.makedirs(output_dir_day8, exist_ok=True)

# Define patch extraction parameters
PATCH_SIZE = 7
HALF_PATCH = PATCH_SIZE // 2

print("TERRASPECTRA - DAY 8: SPATIALLY DISJOINT DATASETS")
print("=" * 60)

# We will split along the x-coordinate (columns) to slice the horizontal leaves into disjoint spatial zones.
# Leaf width is 1024 pixels. We define:
# Train: x_coord in [0, 700]
# Val:   x_coord in [701, 850]
# Test:  x_coord in [851, 1024]

class HyperspectralPatchDataset(Dataset):
    def __init__(self, cube, refined_mask, split="train", patch_size=7):
        self.cube = cube
        self.refined_mask = refined_mask
        self.patch_size = patch_size
        self.half_patch = patch_size // 2

        # Padding the cube to handle edge cases
        self.padded_cube = np.pad(
            cube,
            ((self.half_patch, self.half_patch), (self.half_patch, self.half_patch), (0, 0)),
            mode='reflect'
        )

        # Extract valid coordinates
        all_coords = np.argwhere(refined_mask)
        self.selected_coords = []

        for coord in all_coords:
            y_coord, x_coord = coord[0], coord[1]
            if split == "train":
                if x_coord <= 700:
                    self.selected_coords.append((y_coord, x_coord))
            elif split == "val":
                if 701 <= x_coord <= 850:
                    self.selected_coords.append((y_coord, x_coord))
            elif split == "test":
                if x_coord >= 851:
                    self.selected_coords.append((y_coord, x_coord))

        self.selected_coords = np.array(self.selected_coords)

    def __len__(self):
        return len(self.selected_coords)

    def __getitem__(self, idx):
        y_orig, x_orig = self.selected_coords[idx]

        # Map original coordinates to padded coordinate space
        y_pad = y_orig + self.half_patch
        x_pad = x_orig + self.half_patch

        # Extract spatial-spectral patch
        patch = self.padded_cube[
            y_pad - self.half_patch : y_pad + self.half_patch + 1,
            x_pad - self.half_patch : x_pad + self.half_patch + 1,
            :
        ]

        # Convert to Tensor of shape (Channels/Bands, Height, Width)
        patch_tensor = torch.from_numpy(patch).permute(2, 0, 1).float()

        # Proxy label creation (Upper half vs Lower half)
        label = 1 if y_orig > 250 else 0

        return patch_tensor, torch.tensor(label, dtype=torch.long)

# Instantiate datasets
train_dataset = HyperspectralPatchDataset(roi_cube, refined_mask, split="train", patch_size=PATCH_SIZE)
val_dataset = HyperspectralPatchDataset(roi_cube, refined_mask, split="val", patch_size=PATCH_SIZE)
test_dataset = HyperspectralPatchDataset(roi_cube, refined_mask, split="test", patch_size=PATCH_SIZE)

print("Spatially isolated Dataset counts:")
print(f"Train set size: {len(train_dataset)}")
print(f"Val set size:   {len(val_dataset)}")
print(f"Test set size:  {len(test_dataset)}")
```

```python
import torch.nn as nn
import torch.nn.functional as F

class HyperspectralCNN2D(nn.Module):
    def __init__(self, in_channels=216, patch_size=7, num_classes=2):
        super(HyperspectralCNN2D, self).__init__()

        # Define 2D convolutional layers to extract spatial-spectral features
        self.conv1 = nn.Conv2d(in_channels=in_channels, out_channels=64, kernel_size=3, padding=1)
        self.bn1 = nn.BatchNorm2d(64)

        self.conv2 = nn.Conv2d(in_channels=64, out_channels=32, kernel_size=3, padding=1)
        self.bn2 = nn.BatchNorm2d(32)

        # Global Average Pooling to ensure shape-agnostic processing
        self.gap = nn.AdaptiveAvgPool2d((1, 1))

        self.fc1 = nn.Linear(32, 16)
        self.fc2 = nn.Linear(16, num_classes)

    def forward(self, x):
        x = F.relu(self.bn1(self.conv1(x)))
        x = F.relu(self.bn2(self.conv2(x)))
        x = self.gap(x)
        x = torch.flatten(x, 1)
        x = F.relu(self.fc1(x))
        x = self.fc2(x)
        return x

# Instantiate model and verify forward pass dimensions
model = HyperspectralCNN2D(in_channels=216, patch_size=PATCH_SIZE, num_classes=2)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)

print("TERRASPECTRA - MODEL SUMMARY & DIMENSION CHECK")
print("=" * 60)
test_input = torch.randn(5, 216, PATCH_SIZE, PATCH_SIZE).to(device)
with torch.no_grad():
    test_output = model(test_input)
print("Input dimension (Batch, Bands, H, W):", list(test_input.shape))
print("Output logits dimension (Batch, Classes):", list(test_output.shape))

# Count parameters
total_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Total trainable model parameters: {total_params:,}")
```

```python
import torch.optim as optim
import time

# Setup optimization criteria
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

print("TERRASPECTRA - DAY 8: PROXY MODEL TRAINING")
print("=" * 60)
print("TRAINING ADVISORY: This diagnostic training verifies our PyTorch CNN pipeline structure")
print("using structural proxy labels (upper vs lower leaf halves). Results must NOT be reported as")
print("biological infection accuracy.")
print("-" * 60)

epochs = 5
train_losses, val_losses = [], []
train_accs, val_accs = [], []

for epoch in range(epochs):
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0

    for inputs, labels in train_loader:
        inputs, labels = inputs.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * inputs.size(0)
        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += predicted.eq(labels).sum().item()

    epoch_loss = running_loss / len(train_dataset)
    epoch_acc = correct / total

    # Validation Pass
    model.eval()
    val_running_loss = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():
        for val_inputs, val_labels in val_loader:
            val_inputs, val_labels = val_inputs.to(device), val_labels.to(device)
            val_outputs = model(val_inputs)
            val_loss = criterion(val_outputs, val_labels)

            val_running_loss += val_loss.item() * val_inputs.size(0)
            _, val_predicted = val_outputs.max(1)
            val_total += val_labels.size(0)
            val_correct += val_predicted.eq(val_labels).sum().item()

    val_epoch_loss = val_running_loss / len(val_dataset)
    val_epoch_acc = val_correct / val_total

    train_losses.append(epoch_loss)
    val_losses.append(val_epoch_loss)
    train_accs.append(epoch_acc)
    val_accs.append(val_epoch_acc)

    print(f"Epoch {epoch+1}/{epochs} | "
          f"Train Loss: {epoch_loss:.4f} - Train Acc: {epoch_acc:.4f} | "
          f"Val Loss: {val_epoch_loss:.4f} - Val Acc: {val_epoch_acc:.4f}")
```

```python
# Evaluate model performance on Test Set
model.eval()
test_correct = 0
test_total = 0
all_preds = []
all_targets = []

with torch.no_grad():
    for test_inputs, test_labels in test_loader:
        test_inputs = test_inputs.to(device)
        outputs = model(test_inputs)
        _, predicted = outputs.max(1)
        all_preds.extend(predicted.cpu().numpy())
        all_targets.extend(test_labels.numpy())

all_preds = np.array(all_preds)
all_targets = np.array(all_targets)

# Calculate final metrics
from sklearn.metrics import classification_report, confusion_matrix
print("TERRASPECTRA - MODEL DIAGNOSTIC EVALUATION REPORT")
print("=" * 60)
print("Confusion Matrix:")
print(confusion_matrix(all_targets, all_preds))
print("\nClassification Report:")
print(classification_report(all_targets, all_preds, target_names=["Upper_Half", "Lower_Half"]))
```

#### Theoretical Understanding of 3D Hyperspectral Patches

Unlike 1D spectral vectors (which discard all spatial relationships) or 2D spatial slices (which discard band relationships), a **3D Hyperspectral Patch** ($H \times W \times C$) preserves the local continuity of pixel neighborhoods alongside their full spectral signatures.

When we feed these patches into a 3D CNN, the convolutional kernels slide across height, width, and spectral bands simultaneously. This allows the network to learn unified "spatial-spectral features," such as how local tissue texture interacts with specific water absorption bands.

```python
print("TERRASPECTRA - DAY 9: 3D PATCH EXTRACTION")
print("=" * 60)

# Extraction configuration
PATCH_SIZE = 16
HALF_PATCH = PATCH_SIZE // 2
NUM_BANDS = roi_cube.shape[2]

# Pad the ROI cube to allow extraction around boundary coordinates without out-of-bound crashes
padded_cube = np.pad(
    roi_cube,
    ((HALF_PATCH, HALF_PATCH), (HALF_PATCH, HALF_PATCH), (0, 0)),
    mode='reflect'
)

# To prevent CPU memory exhaustion while demonstrating 3D patch extraction,
# we will extract 3D patches for a representative sample of 1,000 leaf pixels.
np.random.seed(42)
sample_indices = np.random.choice(len(leaf_coordinates_day5), size=1000, replace=False)
sampled_coords = leaf_coordinates_day5[sample_indices]

extracted_patches = []
extracted_coords = []

for (y_orig, x_orig) in sampled_coords:
    # Map original coordinates to padded coordinate space
    y_pad = y_orig + HALF_PATCH
    x_pad = x_orig + HALF_PATCH

    # Slice the 3D patch: (Height, Width, Channels)
    patch = padded_cube[
        y_pad - HALF_PATCH : y_pad + HALF_PATCH,
        x_pad - HALF_PATCH : x_pad + HALF_PATCH,
        :
    ]

    extracted_patches.append(patch)
    extracted_coords.append([y_orig, x_orig])

# Convert to NumPy arrays
hyperspectral_patches = np.array(extracted_patches, dtype=np.float32)
patch_coordinates = np.array(extracted_coords, dtype=np.int64)

print(f"Extracted patches array shape: {hyperspectral_patches.shape}")
print(f"Extracted coordinates array shape: {patch_coordinates.shape}")
```

```python
print("TERRASPECTRA - DAY 9: PATCH VALIDATION")
print("=" * 60)

# 1. Shape Verification
expected_patch_shape = (1000, PATCH_SIZE, PATCH_SIZE, NUM_BANDS)
shape_ok = hyperspectral_patches.shape == expected_patch_shape
print(f"Shape meets expectations ({expected_patch_shape}): {shape_ok}")

# 2. Value Sanity Checks
nan_count = np.isnan(hyperspectral_patches).sum()
inf_count = np.isinf(hyperspectral_patches).sum()
print(f"NaN values: {nan_count}")
print(f"Infinite values: {inf_count}")

# 3. Spatial Correspondence Verification
# Test a random patch to ensure its center matches the raw cube spectral values
test_idx = 42
test_y, test_x = patch_coordinates[test_idx]

# Extract the central pixel spectral signature of the patch
# Since HALF_PATCH = 8, the center pixel of our 16x16 patch is located at indices (8, 8)
patch_center_spectrum = hyperspectral_patches[test_idx, HALF_PATCH, HALF_PATCH, :]
raw_pixel_spectrum = roi_cube[test_y, test_x, :]

correspondence_ok = np.allclose(patch_center_spectrum, raw_pixel_spectrum)
print(f"Central pixel matches raw cube pixel exactly: {correspondence_ok}")

assert shape_ok and nan_count == 0 and inf_count == 0 and correspondence_ok
print("\nSUCCESS: All patch structures and alignment validation tests passed!")
```
