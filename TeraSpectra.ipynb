```python
!pip -q install remotezip

from remotezip import RemoteZip
import os

url = "https://entrepot.recherche.data.gouv.fr/api/access/datafile/:persistentId?persistentId=doi:10.57745/RKRD2T"

folder = "/content/drive/MyDrive/TerraSpectra"

files = [
    "20210331/13_G/13_G.hyspex",
    "20210331/13_G/13_G.hdr"
]

with RemoteZip(url) as z:
    for file in files:
        filename = os.path.basename(file)
        output = os.path.join(folder, filename)

        with z.open(file) as source, open(output, "wb") as target:
            while True:
                chunk = source.read(8 * 1024 * 1024)
                if not chunk:
                    break
                target.write(chunk)

        print("Downloaded:", filename)
```

```python
import os
import spectral

folder = "/content/drive/MyDrive/TerraSpectra"

hdr_path = os.path.join(folder, "13_G.hdr")
img_path = os.path.join(folder, "13_G.hyspex")

img = spectral.envi.open(hdr_path, img_path)

print("Image loaded:", img.shape)
```

```python
import numpy as np
import matplotlib.pyplot as plt

rgb = np.asarray(img[:, :, [80, 52, 15]], dtype=np.float32)

rgb = (rgb - rgb.min()) / (rgb.max() - rgb.min())

plt.figure(figsize=(10, 6))
plt.imshow(rgb)
plt.title("TerraSpectra - VNIR False Colour Image")
plt.axis("off")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

rgb = np.asarray(img[:, :, [80, 52, 15]], dtype=np.float32)

# Improve contrast channel by channel
for i in range(3):
    low, high = np.percentile(rgb[:, :, i], [2, 98])
    rgb[:, :, i] = np.clip(
        (rgb[:, :, i] - low) / (high - low), 0, 1
    )

plt.figure(figsize=(10, 7))
plt.imshow(rgb)
plt.title("TerraSpectra - Enhanced VNIR Image")
plt.axis("off")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Approximate pixel on the upper leaf
row = 550
col = 500

spectrum = img.read_pixel(row, col)

plt.figure(figsize=(10, 5))
plt.plot(spectrum)
plt.title("VNIR Spectral Signature of a Wheat Leaf Pixel")
plt.xlabel("Spectral Band")
plt.ylabel("Pixel Intensity")
plt.grid(True)
plt.show()
```
